```python
df.columns=df.columns.str.strip()
print(df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.stem(df['Total_Engines'],df['Harmless_Votes'])
plt.xlabel("total _Engines")
plt.ylabel("harmless_votes")
plt.title("security _ threats")
plt.show()
```

```python
threats=[df['Total_Engines'],df['Harmless_Votes']]
plt.boxplot(threats,labels=['total_Engines','Harmless_Votes'])
plt.xlabel("total _Engines")
plt.ylabel("harmless_votes")
plt.title("security _ threats")
plt.show()
```
